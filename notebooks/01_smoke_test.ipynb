```python
from pyspark.sql import functions as F
 
df = spark.range(0, 10_000_000)
 
df_filtered = (
    df
    .withColumn("is_even", F.col("id") % 2 == 0)
    .filter(F.col("is_even"))
)
 
print("Transformations defined. Nothing has run yet.")
```

```python
row_count = df_filtered.count()
print(f"Rows after filter: {row_count:,}")
```
