```sql
CREATE CATALOG IF NOT EXISTS sec_dev
COMMENT 'SEC Disclosure Integrity Platform. Source: SEC Financial
Statement Data Sets. Owner: Atif Memon.'
```

```sql
USE CATALOG sec_dev;
```

```sql
CREATE SCHEMA IF NOT EXISTS landing
COMMENT 'Raw unzipped SEC files. Volumes only, no tables.';
 
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Source data as loaded. Append only. One table per SEC source file.';
 
CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Validated and conformed data, plus quarantine and data quality results.';
 
CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Dimensional model and marts. The only layer consumed by BI tools.';
```

```sql
CREATE VOLUME IF NOT EXISTS sec_dev.landing.raw
COMMENT 'Unzipped SEC quarterly TSV files. One subfolder per quarter, for example 2026q1/.';
```
